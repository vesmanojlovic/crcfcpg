```python
import matplotlib.pyplot as plt
from itertools import combinations
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
tumour_label = 'E'
path = '../data/tumour_' + tumour_label + '.csv'
df = pd.read_csv(path)
```

```python
for column in df.columns[1:]:
    plt.figure(figsize=(10, 10))
    df[column].hist(bins=100)
    plt.title(f'Histogram of gland {column} for tumour {tumour_label}')
    plt.xlabel('Beta value')
    plt.ylabel('Frequency')
    plt.show()
```

```python
fig, axs = plt.subplots(8, 8, figsize=(20, 20))

for i in range(8):
    for j in range(8):
        ax = axs[i, j]
        if i < j:  # Scatter plot above the diagonal
            ax.scatter(df.iloc[:, i+1], df.iloc[:, j+1], alpha=0.5)
        elif i > j:  # 2D histogram below the diagonal
            ax.hist2d(df.iloc[:, j+1], df.iloc[:, i+1], bins=30, cmap='Blues')
            ax.set_xticks([])
            ax.set_yticks([])
        else:  # Clear the diagonal plots
            ax.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
correlation_matrix = df.iloc[:, 1:].corr()  # Exclude the labels column
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
